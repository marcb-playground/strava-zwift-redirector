# Repository and `tests/data` Inspection

This notebook inspects the repository root and the `strava-zwift-redirector/tests/data` directory contents.

```python
import os
from pathlib import Path

root = Path('/workspaces/strava-zwift-redirector')
print('Repository root:', root)
print('Exists:', root.exists())
print('Type:', 'directory' if root.is_dir() else 'file')
```

```python
print('\nRepository files and directories:')
for entry in sorted(root.iterdir()):
    print('-', entry.name)
```

```python
data_dir = root / 'strava-zwift-redirector' / 'tests' / 'data'
print('\nInspecting tests/data:')
print('tests/data exists:', data_dir.exists())
print('tests/data is directory:', data_dir.is_dir())
if data_dir.exists():
    for entry in sorted(data_dir.iterdir()):
        print('-', entry.name)
```

```python
gitkeep = data_dir / '.gitkeep'
print('\n.gitkeep presence:')
print('exists:', gitkeep.exists())
print('is file:', gitkeep.is_file())
print('absolute path:', gitkeep.resolve() if gitkeep.exists() else 'N/A')
```

```python
print('\nFile metadata for tests/data contents:')
if data_dir.exists():
    for entry in sorted(data_dir.iterdir()):
        info = entry.stat()
        print(f"{entry.name}: size={info.st_size}, mtime={info.st_mtime}, mode={oct(info.st_mode)}")
```
